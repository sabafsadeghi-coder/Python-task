```python
# 1- Discount Calculator:

product_name = input("Enter the product name:")
product_original_price = int(input("enter the product original price:"))

Discount = product_original_price * 10/100
Final_price = product_original_price - Discount

print(Final_price)
print(product_name)
```

```python
# 2- Student Marks Book:

students = {}

choice = input("Enter your choice: ")

if choice == "1":
    students_name = input("Enter the students name:")
    students_mark = int(input("Enter the students mark:"))

    students[students_name] = students_mark

elif choice == "2":
    name = input("Enter student name: ")

    if name in students:
        print(students[name])
    else:
      print("Student not found")
```

```python
# 3- Restaurant Bill:

customer_name = input("Enter your name: ")
customer_bill_amount = float(input("Enter your bill amount: "))

if customer_bill_amount < 1000:
    discount = 0

elif customer_bill_amount < 3000:
    discount = customer_bill_amount * 5 / 100

elif customer_bill_amount < 5000:
    discount = customer_bill_amount * 10 / 100

else:
    discount = customer_bill_amount * 15 / 100


final_bill = customer_bill_amount - discount

print("Customer:", customer_name)
print("Discount:", discount)
print("Final Bill:", final_bill)
```

```python
# 4- Student Result:

student_name = input("Enter the student name: ")
english = float(input("Enter english marks: "))
math = float(input("Enter mathematics marks: "))
python = float(input("Enter python marks: "))

total = english + math + python
average = total / 3

if average >= 80:
    print("Grade: A")

elif average >= 70:
    print("Grade: B")

elif average >= 60:
    print("Grade: C")

elif average >= 50:
    print("Grade: D")

else:
    print("Grade: Fail")

print("Student name:", student_name)
print("Total Marks:", total)
print("Average:", average)
```

```python
# 5- Hotel Room Booking:

customer_name = input("Enter the customer name: ")
number_of_nights = int(input("Enter the number of nights: "))

if number_of_nights < 3:
    price_per_night = 3000

elif number_of_nights <= 5:
    price_per_night = 2500

elif number_of_nights <= 10:
    price_per_night = 2000

else:
    price_per_night = 1500

total_cost = number_of_nights * price_per_night

print("Customer name:", customer_name)
print("Number of nights:", number_of_nights)
print("Price per night:", price_per_night)
print("Total cost:", total_cost)
```

```python
# 6- Restaurent Order Bill:

total_bill = 0
total_order = int(input("Enter the total orders: "))

for i in range(1, total_order+1):
    order_name = input(f"Enter the name of order  {i}: ")
    price = float(input(f"Enter the price of {order_name}: "))
    quantity = int(input(f"Enter the quantity of {order_name}: "))

    subtotal = price * quantity
    total_bill = total_bill + subtotal

if total_bill >= 3000:
    discount = total_bill * 0.15
else:
    discount = 0

final_bill = total_bill - discount

print(f"Total Bill: Rs. {total_bill:.2f}")
print(f"Discount: Rs. {discount:.2f}")
print(f"Final Bill: Rs. {final_bill:.2f}")
```

```python
# 7- Online Delivery Charge:

delivery_charge = 0

customer_name = input("Enter customer name: ")
order_amount = float(input("Enter the order amount: "))

if order_amount < 1000:
    delivery_charge = 200
elif order_amount < 3000:
    delivery_charge = 150
elif order_amount < 5000:
    delivery_charge = 100
else:
    delivery_charge = "Free Delivery"

print("Customer name:", customer_name)
print("Order amount:", order_amount)
print("Delivery charge:", delivery_charge)
```

```python
# 8- Temperture Checker:

temperature = float(input("Enter the temperature: "))

if temperature > 30:
    print("Weather: Hot")

elif temperature >= 20:
    print("Weather: Warm")

elif temperature >= 10:
    print("Weather: Cool")

else:
    print("Weather: Cold")
```
