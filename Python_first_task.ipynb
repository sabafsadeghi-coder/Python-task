```python
#Even or Odd
number = int(input("Enter a number: "))

if number % 2 == 0:
    print("Even")

else:
    print("Odd")
```

```python
#simple calculator
num1 = float(input("enter then first number: "))
num2 = float(input("enter the second number: "))
operator = input("enter an operator (+ , - , * , / ): ")

if operator == "+":
   result = num1 + num2
elif operator == "-":
   result = num1 - num2
elif operator == "*":
   result = num1 * num2
elif operator == "/":
   result = num1 / num2
else:
   result = "invalid operator"

print("Result:", result)
```

```python
#student marks
mark1 = float(input("enter marks of subject 1: "))
mark2 = float(input("enter marks of subject 2: "))
mark3 = float(input("enter marks of subject 3: "))

total = mark1 + mark2 + mark3
percentage = (total / 300 * 100)

print("Total:" , total )
print("Percentage:" , percentage )

if percentage >= 50:
     print("Pass")

else:
   print("Fail")
```

```python
#Favorite fruits
fruits = ["orange" , "banana" , "watermelon" , "pineapple" , "mango"]

fruit = input("enter a fruit: ")

if fruit in fruits:
   print("fruit exists in the list")
else:
   print("fruit does not exist in the list")
```
