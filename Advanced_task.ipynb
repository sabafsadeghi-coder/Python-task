```python
# Task 11: Personal Expense tracker

expenses = []

total_expenses = int(input("How many expenses do you want to enter? "))

for i in range(total_expenses):
    print("Enter expenses", i + 1)
    name = input("Enter expense name: ")
    category = input("Enter category: ")
    amount = float(input("Enter amount: "))

    expense = {
        "name": name,
        "category": category,
        "amount": amount
    }

    expenses.append(expense)

total = 0

for expense in expenses:
    total = total + expense["amount"]

print("Total Spending:", total)

category_spending = {

    "Food": 0,
    "Transport": 0,
    "Shopping": 0,
    "Bills": 0,
    "Other": 0
}
for expense in expenses:
    category = expense["category"]
    amount = expense["amount"]

    category_spending[category] += amount

highest = expenses[0]

for expense in expenses:
    if expense["amount"] > highest["amount"]:
      highest = expense

print("Highest Expense:", highest["name"])
print("Amount:", highest["amount"])

budget = float(input("Enter your budget: "))

if total > budget:
    print("Warning: Budget exceeded!")
else:
    print("Your are within your budget.")

def display_report():
    print("\n========== EXPENSE REPORT ==========")
    print("Total Spending:", total)

    print("\nCategory Spending:")
    print("Food:", category_spending["Food"])
    print("Transport:", category_spending["Transport"])
    print("Shopping:", category_spending["Shopping"])
    print("Bills:", category_spending["Bills"])
    print("Other:", category_spending["Other"])

    print("\nHighest Expense:", highest["name"])
    print("Amount:", highest["amount"])

    print("\nBudget:", budget)

    if total > budget:
        print("Warning: Budget exceeded!")
    else:
        print("You are within your budget.")

display_report()
```

```python
# Task 12: Store Inventory Manager

products = {
    "Milk": {
        "price": 200,
        "quantity": 10
    },
    "Bread": {
        "price": 150,
        "quantity": 4
    }

}


def add_product():
    name = input("Enter product name:")
    price = float(input("Enter product price:"))
    quantity = int(input("Enter product quantity:"))

    if name in products:
        print("Product already exists")
        return

    products[name] = {
        "price": price,
        "quantity": quantity
    }

    print("Product added successfully")


def sell_product():
    name = input("Enter product name: ")
    quantity = int(input("Enter quantity to sell: "))

    if name not in products:
        print("Product not found")
        return

    if products[name]["quantity"] < quantity:
        print("Insufficient stock")
        return

    products[name]["quantity"] -= quantity
    print("Product sold successfully")


def restock_product():
    name = input("Enter product name: ")
    quantity = int(input("Enter quantity to restock: "))

    if name not in products:
        print("Product not found")
        return

    products[name]["quantity"] += quantity
    print("Product restocked successfully")


def search_product():
    name = input("Enter product name: ")

    if name in products:
        print(products[name])
    else:
        print("Product not found")


def display_inventory():
    total_value = 0

    for name, product in products.items():
        print(name, "price:", product["price"],
              "Quantity:", product["quantity"])

        if product["quantity"] < 5:
            print("low stock: ", name)

        total_value += product["price"] * product["quantity"]
    print("Total inventory value: ", total_value)


while True:
    print("1. Add product")
    print("2. Sell product")
    print("3. Restock product")
    print("4. Search product")
    print("5. Display inventory")
    print("6. Exit")

    choice = int(input("Enter your choice: "))
    if choice == 1:
        add_product()
    elif choice == 2:
        sell_product()
    elif choice == 3:
        restock_product()
    elif choice == 4:
        search_product()
    elif choice == 5:
        display_inventory()
    elif choice == 6:
        print("Program exited successfully.")
        break

    else:
        print("Invalid choice")
```

```python
# Task 13: Library Book Management System

books = []

def add_book():
    book_id = int(input("Enter book ID: "))
    title = input("Enter book title: ")
    author = input("Enter author name: ")

    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists")
            return

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)

    print("Book added successfully")


def display_books():
    for book in books:
        print("ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Available:", book["available"])


def search_book():
    title = input("Enter book title: ")

    for book in books:
        if book["title"] == title:
            print(book)
            return
    print("Book not found")


def borrow_book():
    book_id = int(input("Enter book ID"))

    for book in books:
        if book["id"] == book_id:
            if not book["available"]:
                print("Book is already borrowed")
                return

            book["available"] = False
            print("Book borrowed successfully")
            return

    print("Book not found")


def return_book():
    book_id = int(input("Enter book ID: "))

    for book in books:
        if book["id"] == book_id:
            book["available"] = True
            print("Book returned successfully")
            return
    print("Book not found")


while True:
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        add_book()
    elif choice == 2:
        display_books()
    elif choice == 3:
        search_book()
    elif choice == 4:
        borrow_book()
    elif choice == 5:
        return_book()
    elif choice == 6:
        print("Program exited successfully.")
        break
    else:
        print("Invalid choice")
```

```python
# Task 14: Restaurant Ordering System

menu ={
     "Burger":500,
     "Pizza":1200,
     "Pasta":800,
     "Drink":200
}
orders = []

print("======== Restaurant Menu ========")
for key,value in menu.items():
    print(f'{key}:{value}')

while True:
  item_name = input("Enter your item name: ").title()
  if item_name in menu:
    quantity = int(input("Enter quantity of your item: "))
    price = menu[item_name]
    sub_total = price * quantity

    orders.append({
        "item": item_name,
        "quantity": quantity,
        "sub_total": sub_total
    })
  else:
     print("This item is not available")

  choice = input("You want to add more items? (yes/no): ")
  if choice == "no":
       break



total = 0

for order in orders:
    total = total + order["sub_total"]

tax = total  * 0.13
# tax = (total * 13) / 100

if total >= 3000:
    discount = total * 0.1
else:
    discount = 0

grand_total = total + tax - discount

def receipt():
    print("\n======= Receipt =======")

    for order in orders:
      print(f'{order["item"]} x {order["quantity"]} = {order["sub_total"]}')

    print("Total: ", total)
    print("Tax: ", tax)
    print("Discount: ", discount)
    print("Grand Total: ", grand_total)
    #print("========== After discount ==========")
    #print("final_bill: ", grand_total + discount)

receipt()
```

```python
# Task 15: Clinic Appointment Manager
patients = []

def add_patient():
  patient_id = int(input("Enter your id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient already exists")
      return
  patient_name = input("Enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("Enter patient problem: ")
  time = int(input("Enter appointment time: "))
  emergency = input("Is emergency pateint? yes/no: ")

  patients.append({
    "id": patient_id,
    "name": patient_name,
    "age": patient_age,
    "problem": problem,
    "time": time,
    "emergency": emergency
  })

def search_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      print("Patient found")
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')
      print(f'Patient emergency: {patient["emergency"]}')
      return
  else:
   print("Patient not found")

def display_appointment():
  print("========== Emergency pateint ==========")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("========== normal pateint ==========")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

def cancel_appointment():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["id"] == patient_id:
      patient["time"] = "cancelled"
      print("Appointment canceled")
      return


def count_patient():
  emergency_count = 0
  normal_count = 0

  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1

  print(f'Total emergency pateint: {emergency_count}')
  print(f'Total normal pateint: {normal_count}')

while True:
  print("1. Add patient")
  print("2. Search patient")
  print("3. Display appointment")
  print("4. Cancel patient")
  print("5. Count patient")
  print("6. Exit")

  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_patient()
  elif choice == 2:
    search_patient()
  elif choice == 3:
    display_appointment()
  elif choice == 4:
    cancel_appointment()
  elif choice == 5:
    count_patient()
  elif choice == 6:
    break
  else:
    print("Invalid choice")
```
